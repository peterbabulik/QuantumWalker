<a name="the-core-idea"></a>
## 1. The Core Idea: Inventing Tools, Not Just Solutions

Most machine learning for quantum computing focuses on finding the parameters for a *fixed* algorithm (ansatz) to solve a *specific* problem. **CMA-ES QGF** takes a more fundamental approach:

> Can an AI, given only the basic laws of quantum mechanics, invent a new, universal building block (a quantum gate) that is powerful for a wide range of computational tasks?

Instead of being a problem-solver, the AI becomes a **toolmaker**. We hypothesize that by optimizing for a fundamental, problem-agnostic physical property—**maximal entangling power**—the AI will discover gates that are inherently useful for computation.

<a name="the-experimental-architecture"></a>
## 2. The Experimental Architecture: A Two-Act Play

The main script (`quantum_forge.py`) is structured as a two-act play that runs in a loop to gather robust statistics.

<a name="act-1-the-forge"></a>
### Act 1: The Forge

*   **Objective:** Create a novel 2-qubit unitary gate `U` that is a **maximally powerful entangler**.
*   **AI Method:** We use CMA-ES to evolve the **15 real-valued coefficients** of a gate's **Generator Hamiltonian `H`**. The final gate is then constructed via the Schrödinger equation: `U = e^(-iH)`. This "Generator Method" guarantees the resulting gate is always unitary and provides a physically meaningful "DNA" for the AI to evolve.
*   **Fitness Function:** The "quality" of a forged gate is measured by the **Meyer-Wallach entangling power `Q(U)`**. This metric ranges from 0 (for non-entangling gates) to 1 (for perfect entanglers like CNOT). The AI's fitness function is `1.0 - Q(U)`, which it seeks to minimize to zero.

<a name="act-2-the-test-drive"></a>
### Act 2: The Test Drive

*   **Objective:** Test the computational utility of the newly forged gate (`FORGED_U`).
*   **The Task:** We challenge the gate to solve a genuinely hard, entangled problem: finding the ground state of the 5-qubit Max-Cut Hamiltonian for a complete graph (K5).
*   **The Ansatz:** A simple, fixed circuit structure `['RY', 'FORGED_U', 'RY']` is used. This isolates the performance of the forged gate itself.
*   **Evaluation:** A *second* CMA-ES optimization finds the best rotation angles for the `RY` gates. The final performance is measured by the **state fidelity** between the ansatz's output state and the true, pre-computed ground state of the problem. A fidelity of 100% means the gate enabled a perfect solution.

<a name="key-discoveries-and-the-journey"></a>
## 3. Key Discoveries and The Journey

This project was not a straight line to success. The AI's "failures" and "lazy" answers were crucial in debugging the experimental setup and revealing deeper truths:

1.  **Exposing Flawed Metrics:** In early stages, the AI found "loopholes" in simple fitness functions, forcing the development of more robust metrics like direct state fidelity.
2.  **Uncovering Hamiltonian Bugs:** The AI's consistent refusal to produce an entangled state helped uncover a critical sign error in the problem Hamiltonian, revealing we were accidentally asking it to solve a simple problem, not a hard one.
3.  **Demonstrating Robustness:** The final multi-run experiment shows that the CMA-ES QGF framework is not a fluke. It **consistently** discovers maximally entangling gates that are **consistently** useful for solving a hard computational problem.
4.  **The Main Conclusion:** We have experimentally demonstrated that **maximal entangling power is an excellent heuristic for designing computationally powerful quantum gates.** The AI, optimizing for this abstract physical property, created tools that were immediately effective for practical computation.

````python

# ==============================================================================
#  PREAMBLE: IMPORTS AND SETUP
# ==============================================================================
import numpy as np
import time
from typing import List, Dict, Tuple
import networkx as nx
from scipy.linalg import expm
import warnings

# Qiskit Imports
import qiskit
from qiskit.circuit import QuantumCircuit, ParameterVector, Gate
from qiskit.quantum_info import SparsePauliOp, Operator, Statevector, partial_trace
from qiskit_aer import AerSimulator
# Import the specific synthesis tools we need for Qiskit 2.x
from qiskit.synthesis.two_qubit import TwoQubitBasisDecomposer
from qiskit.circuit.library import CXGate

# The powerful "Designer AI" engine
import cma

# Suppress benign warnings from CMA-ES and NumPy for cleaner output
warnings.filterwarnings("ignore", category=RuntimeWarning)
warnings.filterwarnings("ignore", category=FutureWarning)


print(f"Qiskit version: {qiskit.__version__}")
print("\n--- The Quantum Gate Forge: Multi-Run Experiment (Definitive Version) ---")
print("We will forge and test 10 unique gates to find the best tool.")

# ==============================================================================
#  CORE FUNCTIONS (UNCHANGED)
# ==============================================================================

PAULI_BASIS_2Q = [p1+p2 for p1 in 'IXYZ' for p2 in 'IXYZ' if p1+p2 != 'II']

def build_unitary_from_generator(coeffs: np.ndarray) -> np.ndarray:
    """Constructs a 2Q unitary matrix from the 15 coeffs of its generator H."""
    coeffs = np.asarray(coeffs).flatten()
    generator_h = SparsePauliOp(PAULI_BASIS_2Q, coeffs=coeffs)
    h_matrix = generator_h.to_matrix()
    u_matrix = expm(-1j * h_matrix)
    return u_matrix

def calculate_entangling_power(u_matrix: np.ndarray) -> float:
    """Calculates the Meyer-Wallach entangling power Q(U) of a 2Q gate."""
    psi = np.zeros(4, dtype=np.complex128)
    psi[0] = 1/np.sqrt(2)
    psi[3] = 1/np.sqrt(2)
    psi_evolved = u_matrix @ psi
    rho_A = partial_trace(psi_evolved, [1]).data
    purity = np.real(np.trace(rho_A @ rho_A))
    entangling_power = 2 * (1 - purity)
    return entangling_power

# ==============================================================================
#  MAIN EXPERIMENT LOOP
# ==============================================================================

if __name__ == "__main__":

    NUM_RUNS = 10
    results_log = []
    best_run_data = {'fidelity': -1.0}

    # --- Setup the Max-Cut problem (we only need to do this once) ---
    g_hard = nx.complete_graph(5)

    def get_correct_max_cut_hamiltonian(graph: nx.Graph) -> SparsePauliOp:
        pauli_list = []
        num_nodes = graph.number_of_nodes()
        for i, j in graph.edges():
            p_str = ['I'] * num_nodes
            p_str[i], p_str[j] = 'Z', 'Z'
            pauli_list.append("".join(p_str))
        return SparsePauliOp(pauli_list, coeffs=np.ones(len(pauli_list)))

    correct_hamiltonian = get_correct_max_cut_hamiltonian(g_hard)
    h_matrix = correct_hamiltonian.to_matrix()
    eigvals, eigvecs = np.linalg.eigh(h_matrix)
    true_entangled_ground_state = eigvecs[:, 0]


    # --- Main Loop: Run the entire experiment 10 times ---
    for run in range(NUM_RUNS):
        print(f"\n{'='*60}\n--- RUN {run + 1}/{NUM_RUNS} ---\n{'='*60}")

        # --- ACT 1: FORGE A NEW GATE ---
        print(f"\n[RUN {run+1}] Act 1: Forging a new gate...")

        NUM_GENERATOR_COEFFS = 15
        x0_forge = np.random.rand(NUM_GENERATOR_COEFFS) * 2 * np.pi - np.pi
        sigma0_forge = 0.5
        options_forge = {'bounds': [-np.pi, np.pi], 'maxfevals': 2000, 'verbose': -9}
        es_forge = cma.CMAEvolutionStrategy(x0_forge, sigma0_forge, options_forge)

        def forge_fitness(coeffs: np.ndarray) -> float:
            try:
                u_matrix = build_unitary_from_generator(coeffs)
                return 1.0 - calculate_entangling_power(u_matrix)
            except Exception: return 1.0

        es_forge.optimize(forge_fitness)

        best_coeffs = es_forge.result.xbest
        forged_u_matrix = build_unitary_from_generator(best_coeffs)
        forged_power = calculate_entangling_power(forged_u_matrix)

        print(f"  > Forge complete. Entangling Power Q(U): {forged_power:.4f}")

        # --- Synthesize the forged gate ---
        FORGED_GATE = Gate(name=f"FORGED_{run+1}", num_qubits=2, params=[])
        decomposer = TwoQubitBasisDecomposer(CXGate())
        synthesis_result = decomposer(forged_u_matrix)

        # --- THE FINAL FIX IS HERE ---
        # The result of the decomposer IS the circuit definition
        FORGED_GATE.definition = synthesis_result
        # --- END OF FIX ---

        # --- ACT 2: TEST THE NEW GATE ---
        print(f"[RUN {run+1}] Act 2: Test driving the new gate...")

        def create_test_drive_ansatz(num_qubits: int):
            qc = QuantumCircuit(num_qubits)
            params = ParameterVector('θ', num_qubits * 2)
            for i in range(num_qubits): qc.ry(params[i], i)
            for i in range(num_qubits - 1): qc.append(FORGED_GATE, [i, i+1])
            for i in range(num_qubits): qc.ry(params[num_qubits + i], i)
            return qc

        test_ansatz = create_test_drive_ansatz(5)

        def run_vqe_and_get_fidelity(ansatz, params):
            bound_circuit = ansatz.assign_parameters(params)
            evolved_state = Statevector.from_instruction(bound_circuit).data
            return qiskit.quantum_info.state_fidelity(true_entangled_ground_state, evolved_state)

        def test_drive_fitness(ry_params: np.ndarray) -> float:
            return 1.0 - run_vqe_and_get_fidelity(test_ansatz, ry_params)

        x0_test = np.random.rand(test_ansatz.num_parameters) * 2 * np.pi
        sigma0_test = 0.5
        options_test = {'bounds': [0, 2*np.pi], 'maxfevals': 1000, 'verbose': -9}
        es_test = cma.CMAEvolutionStrategy(x0_test, sigma0_test, options_test)

        es_test.optimize(test_drive_fitness)
        final_fidelity = 1.0 - es_test.result.fbest
        print(f"  > Test drive complete. Final Fidelity: {final_fidelity:.2%}")

        # --- Log the results for this run ---
        top_coeffs = sorted(zip(PAULI_BASIS_2Q, best_coeffs), key=lambda item: abs(item[1]), reverse=True)
        run_summary = {
            'run': run + 1,
            'power': forged_power,
            'fidelity': final_fidelity,
            'coeffs': top_coeffs[:5] # Store the top 5 "genes"
        }
        results_log.append(run_summary)

        # Check if this is the best run so far
        if final_fidelity > best_run_data['fidelity']:
            best_run_data = run_summary


    # ==============================================================================
    #  FINAL SUMMARY AND ANALYSIS
    # ==============================================================================
    print(f"\n\n{'='*60}\n--- OVERALL EXPERIMENT SUMMARY ---")
    print(f"{'='*60}\n")

    print("Run-by-Run Performance:")
    print("-" * 50)
    print(f"{'Run':<5} | {'Entangling Power':<20} | {'Final Fidelity':<15}")
    print("-" * 50)

    fidelities = []
    for res in results_log:
        fidelities.append(res['fidelity'])
        print(f"{res['run']:<5} | {res['power']:.4f}{'':<14} | {res['fidelity']:.2%}")
    print("-" * 50)

    avg_fidelity = np.mean(fidelities)
    max_fidelity = np.max(fidelities)

    print(f"\nAverage Fidelity Achieved: {avg_fidelity:.2%}")
    print(f"Best Fidelity Achieved:    {max_fidelity:.2%}")

    print(f"\n--- Analysis of the Champion Gate (from Run {best_run_data['run']}) ---")
    print(f"This gate achieved the highest fidelity of {best_run_data['fidelity']:.2%}.")
    print("Its generator H = sum(c_i * P_i) was defined by:")
    for pauli, coeff in best_run_data['coeffs']:
        print(f"  {pauli}: {coeff:.3f}")

    if max_fidelity > 0.95:
        print("\nOverall Conclusion: SUCCESS! The AI consistently forged powerful and useful gates.")
    else:
        print("\nOverall Conclusion: FASCINATING! The results show that while the AI can always")
        print("create maximally entangling gates, their utility depends heavily on the specific")
        print("type of interactions (the 'DNA') they are built from.")
````
