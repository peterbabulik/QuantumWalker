```python

# ==============================================================================
#  Project Prometheus: The Genesis Engine
#
#  An AI built from first principles to evolve a computational universe.
#
#  Objective: To discover a set of fundamental "laws of physics" (a kernel)
#  that can give rise to stable, complex, emergent structures from a
#  random initial state.
# ==============================================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from scipy.signal import convolve2d
from tqdm import tqdm
from IPython.display import HTML

# ==============================================================================
#  The Universe Class: The Substrate and its Simulation
# ==============================================================================
class Universe:
    def __init__(self, size=128):
        self.size = size
        # Initialize the substrate with random "quantum foam"
        self.grid = np.random.rand(size, size)

    def run_simulation(self, kernel: np.ndarray, steps: int):
        """Evolves the universe according to its laws of physics (the kernel)."""
        history = [self.grid.copy()]
        for _ in range(steps):
            # The core of the physics: a local convolution
            self.grid = convolve2d(self.grid, kernel, mode='same', boundary='wrap')
            # A non-linear activation function (e.g., tanh) to keep potentials bounded
            self.grid = np.tanh(self.grid)
            history.append(self.grid.copy())
        return history

# ==============================================================================
#  The Prometheus AI: The Meta-Algorithm
# ==============================================================================
class Prometheus:
    def __init__(self, population_size=50, generations=100, mutation_rate=0.1):
        self.population_size = population_size
        self.generations = generations
        self.mutation_rate = mutation_rate

        # The "genes" are the 9 values of a 3x3 kernel
        self.num_genes = 9
        # Initialize a random population of possible laws of physics
        self.population = [np.random.randn(3, 3) for _ in range(population_size)]

        print("Prometheus AI initialized.")
        print(f"Population Size: {self.population_size} universes")
        print(f"Generations: {self.generations}")

    def _calculate_fitness(self, kernel: np.ndarray, sim_steps=50):
        """The fitness function that rewards elegance and fertility."""

        # --- 1. Fertility Score ---
        # A good universe creates stable, non-trivial structures.
        universe = Universe()
        history = universe.run_simulation(kernel, sim_steps)
        final_grid = history[-1]

        # Measure the amount of structure (variance) in the final state.
        # A sterile universe (all zero or all uniform) will have low variance.
        structure_measure = np.var(final_grid)

        # Measure the stability of the structure.
        # A stable universe doesn't change much in the last few steps.
        stability_measure = 1.0 / (1.0 + np.mean(np.abs(history[-1] - history[-2])))

        # A universe that explodes or dies is penalized
        if np.isclose(structure_measure, 0) or np.any(np.isnan(final_grid)):
            return -1e9 # Extremely unfit

        fertility_score = structure_measure * stability_measure

        # --- 2. Elegance Cost ---
        # Simple laws are better. The cost is the sum of the "strengths" of the laws.
        elegance_cost = np.sum(np.abs(kernel))

        # --- Total Fitness ---
        # We want to maximize fertility while minimizing elegance cost.
        total_fitness = fertility_score - (0.01 * elegance_cost)
        return total_fitness

    def evolve(self):
        """The main evolutionary loop (Cosmological Natural Selection)."""
        print("\n--- Beginning Cosmological Evolution ---")
        best_kernel_so_far = None
        best_fitness_so_far = -1e9

        for gen in tqdm(range(self.generations), desc="Evolving Universes"):
            # 1. Evaluate the fitness of each universe in the population
            fitness_scores = [self._calculate_fitness(kernel) for kernel in self.population]

            # 2. Selection: Keep the top 20% of universes
            fittest_indices = np.argsort(fitness_scores)[-int(self.population_size * 0.2):]
            survivors = [self.population[i] for i in fittest_indices]

            if fitness_scores[fittest_indices[-1]] > best_fitness_so_far:
                best_fitness_so_far = fitness_scores[fittest_indices[-1]]
                best_kernel_so_far = survivors[-1]

            # 3. Create the next generation
            next_generation = survivors.copy() # The fittest survive as-is
            while len(next_generation) < self.population_size:
                # Crossover: Create a child from two random survivors
                parent1, parent2 = np.random.choice(len(survivors), 2, replace=False)
                parent1_genes = survivors[parent1].flatten()
                parent2_genes = survivors[parent2].flatten()

                crossover_point = np.random.randint(1, self.num_genes)
                child_genes = np.concatenate((parent1_genes[:crossover_point], parent2_genes[crossover_point:]))

                # Mutation: Apply small random changes to the child's laws
                if np.random.rand() < self.mutation_rate:
                    child_genes += np.random.randn(self.num_genes) * 0.1

                next_generation.append(child_genes.reshape(3, 3))

            self.population = next_generation

        print("\n--- Evolution Complete ---")
        return best_kernel_so_far

    def animate_champion(self, champion_kernel, steps=200):
        """Visualize the universe created by the best laws of physics found."""
        print("\n--- Simulating the Champion Universe ---")
        universe = Universe()
        history = universe.run_simulation(champion_kernel, steps)

        fig, ax = plt.subplots(figsize=(8, 8))
        im = ax.imshow(history[0], cmap='plasma', vmin=-1, vmax=1)
        ax.set_title("The Champion Universe | Time Step: 0")
        ax.axis('off')

        def animate(i):
            im.set_data(history[i])
            ax.set_title(f"The Champion Universe | Time Step: {i}")
            return [im]

        ani = FuncAnimation(fig, animate, frames=len(history), interval=50, blit=True)
        plt.close()
        display(HTML(ani.to_jshtml()))

# ==============================================================================
#  MAIN EXECUTION: Let There Be Light
# ==============================================================================
if __name__ == "__main__":
    # Instantiate the Meta-Algorithm
    ai_god = Prometheus(population_size=100, generations=50, mutation_rate=0.2)

    # Run the evolutionary process to discover the laws of physics
    fittest_laws = ai_god.evolve()

    print("\n--- The Fittest Laws of Physics Discovered ---")
    print("Kernel (The 'Genes' of the Universe):")
    print(np.round(fittest_laws, 4))

    # Simulate and visualize the universe that these laws create
    ai_god.animate_champion(fittest_laws, steps=300)
```
