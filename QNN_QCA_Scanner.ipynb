# The QCA Universe Scanner: Mapping Emergent Complexity

This repository contains the code and results of a systematic computational experiment designed to map the relationship between system size and emergent complexity in a quantum system. It builds upon the initial discovery of "spaceships" in a Quantum Cellular Automaton (QCA) by automating the discovery process across a range of universes.

The project's goal is to move beyond finding a single "lifeform" to understanding the fundamental "laws of biology" that govern these quantum universes.

## The Scientific Question

After discovering that a simple QCA could support complex, self-propagating patterns, a critical question arose: **How does the potential for this complexity change as we change the size of the universe?**

Is the relationship simple and linear, or is it governed by deeper, more subtle properties of the system? To answer this, we built the "Universe Scanner."

## The Experiment: An Automated Science Machine

The "Universe Scanner" is a meta-experiment that systematically performs the following tasks:

1.  **Iterates through Universes:** It loops through a range of system sizes (`N` = number of qubits).
2.  **Launches Discovery Probes:** For each universe size `N`, it runs our QNN-based discovery algorithm hundreds of times with different random seeds. The goal for each run is to find initial states that evolve into stable, periodic patterns (oscillators/spaceships).
3.  **Collects and Curates Data:** The scanner gathers all "perfect" solutions (`cost=0`) found across all runs.
4.  **Identifies "Species":** It processes the raw solutions to identify unique "families of life." It does this by treating all rotated versions of a pattern as the same fundamental species (e.g., `[1,0,0,0]` and `[0,1,0,0]` are considered the same).
5.  **Analyzes the Results:** Finally, it plots the number of unique species discovered as a function of the universe size `N`, revealing the "Law of Complexity" for this system.

## The Results: A Law of Complexity Emerges

The scan produced a fascinating and non-obvious result, which represents the core discovery of this project.

The experimental data allowed us to reject several simple hypotheses:

### The Current Leading Hypothesis:

The complexity of a universe in this QCA appears to be a function of both its **size (`N`)** and its **number-theoretic properties (divisors and symmetries)**. For example, the unique three-fold symmetry of the N=9 universe seems to enable new, resonant, period-3 patterns that are impossible in other systems.

````python

# ==============================================================================
#  PREAMBLE: IMPORTS AND SETUP
# ==============================================================================
import numpy as np
import matplotlib.pyplot as plt
import time
from typing import Dict, List, Tuple, Set

# Qiskit Imports
import qiskit
from qiskit import QuantumCircuit
from qiskit.circuit import ParameterVector
from qiskit.circuit.library import CXGate, CZGate, SwapGate, IGate, XGate
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator

print("--- QCA Universe Scanner ---")
print(f"Qiskit version: {qiskit.__version__}")

# ==============================================================================
#  CORE COMPONENTS (Encapsulated from previous script)
# ==============================================================================

# (The QuantumCellularAutomaton, create_generator_qnn, and
# create_cost_function_with_disqualification functions are identical to the
# previous script. They are included here for completeness.)

GATES = {"CX": CXGate(), "CZ": CZGate(), "SWAP": SwapGate(), "I": IGate()}
class QuantumCellularAutomaton:
    def __init__(self, num_qunodes: int, interaction_config: Dict):
        self.num_qunodes = num_qunodes
        self.interaction_config = interaction_config
        self.simulator = AerSimulator(method='statevector')
    def _get_interaction_unitary(self, t1, t2): return self.interaction_config.get(tuple(sorted((t1, t2))), GATES["I"])
    def evolve_state(self, initial_state: np.ndarray) -> np.ndarray:
        qc = QuantumCircuit(self.num_qunodes)
        for i, bit in enumerate(initial_state):
            if bit == 1: qc.x(i)
        qc.barrier()
        for i in range(self.num_qunodes):
            j = (i + 1) % self.num_qunodes
            op = self._get_interaction_unitary(initial_state[i], initial_state[j])
            if op.name != 'id': qc.append(op, [i, j])
        sv = Statevector(qc)
        probs = sv.probabilities_dict()
        max_prob_str = max(probs, key=probs.get)
        return np.array([int(bit) for bit in max_prob_str[::-1]])

def create_generator_qnn(n_q: int, d: int) -> QuantumCircuit:
    theta = ParameterVector('θ', n_q * d)
    qc = QuantumCircuit(n_q)
    for i in range(d):
        for j in range(n_q): qc.ry(theta[i * n_q + j], j)
        if n_q > 1:
            for j in range(n_q - 1): qc.cz(j, j + 1)
    qc.measure_all()
    return qc

def create_cost_function(qca: QuantumCellularAutomaton, qnn: QuantumCircuit, steps: int, shift: int):
    n_q = qnn.num_qubits
    backend = AerSimulator()
    def cost_func(theta):
        bound_qc = qnn.assign_parameters(theta)
        res = backend.run(bound_qc, shots=1, memory=True).result()
        s_init_str = res.get_memory()[0]
        s_init = np.array([int(b) for b in s_init_str[::-1]])
        activity = np.sum(s_init)
        if activity == 0 or activity == n_q: return n_q, s_init
        s_final = s_init
        for _ in range(steps): s_final = qca.evolve_state(s_final)
        s_target = np.roll(s_init, shift)
        cost = np.sum(s_final != s_target)
        return cost, s_init
    return cost_func

# ==============================================================================
#  THE META-EXPERIMENT: A FUNCTION TO RUN ONE FULL DISCOVERY
# ==============================================================================

def find_solutions_for_config(
    N: int, D: int, S: int, G: int,
    max_iter: int, num_runs: int, rules: Dict
) -> Set[Tuple[int, ...]]:
    """
    Runs the discovery process multiple times for a single configuration
    and returns the set of unique, perfect solutions found.
    """
    print(f"\n--- Scanning N={N}, D={D}, S={S}, G={G} ---")

    qca_universe = QuantumCellularAutomaton(num_qunodes=N, interaction_config=rules)
    generator_qnn = create_generator_qnn(n_q=N, d=D)
    cost_function = create_cost_function(qca=qca_universe, qnn=generator_qnn, steps=S, shift=G)

    unique_solutions_found = set()

    for i in range(num_runs):
        print(f"  Run {i+1}/{num_runs}...", end='\r')
        best_cost = float('inf')

        # We don't need to save the best state, just run the search
        for _ in range(max_iter):
            # Using a simpler search for speed in the meta-experiment
            params = np.random.uniform(0, 2 * np.pi, generator_qnn.num_parameters)
            cost, state = cost_function(params)

            if cost < best_cost:
                best_cost = cost

            if cost == 0:
                # Found a perfect solution! Add it to our set.
                # Convert to tuple to make it hashable for the set.
                unique_solutions_found.add(tuple(state))
                break # Move to the next run

    print(f"\nCompleted scan for N={N}, D={D}. Found {len(unique_solutions_found)} unique solutions.")
    return unique_solutions_found

# ==============================================================================
#  THE MAIN CONTROL SCRIPT
# ==============================================================================

if __name__ == "__main__":
    # --- Define the Scan Parameters ---
    # Start small to test, then expand
    N_range = range(4, 9)  # Test N=4, 5, 6, 7, 8
    D_range = range(4, 13, 4) # Test D=4, 8, 12

    # Fixed parameters for this meta-experiment
    # We are looking for period-4 oscillators (shift=0 for N=4, shift=-4 for N=8)
    QCA_SIM_STEPS = 4
    GLIDER_SHIFT_TEMPLATE = -4 # We'll adjust this based on N

    RUNS_PER_CONFIG = 100
    ITER_PER_RUN = 500  # Max iterations for each of the 100 runs

    # --- Define the "Laws of Physics" ---
    QCA_RULES = {
        (0, 0): IGate(),
        (1, 1): CXGate(),
        (0, 1): SwapGate()
    }

    # --- Data Collection ---
    all_results = {}
    scan_start_time = time.time()

    for n_qubits in N_range:
        for qnn_depth in D_range:
            # The shift must wrap around correctly. A shift of -N is a shift of 0.
            glider_shift = GLIDER_SHIFT_TEMPLATE % n_qubits

            solutions = find_solutions_for_config(
                N=n_qubits, D=qnn_depth, S=QCA_SIM_STEPS, G=glider_shift,
                max_iter=ITER_PER_RUN, num_runs=RUNS_PER_CONFIG, rules=QCA_RULES
            )
            all_results[(n_qubits, qnn_depth)] = solutions

    scan_end_time = time.time()
    print(f"\n--- META-EXPERIMENT COMPLETE IN {scan_end_time - scan_start_time:.2f} SECONDS ---")

    # --- STATISTICAL ANALYSIS AND PREDICTION ---
    print("\n--- Analysis of Results ---")

    # We will analyze the number of solutions as a function of N,
    # taking the result from the deepest (most powerful) QNN for each N.
    num_solutions_per_n = {}
    for n_qubits in N_range:
        max_depth_for_n = max(D_range)
        solutions = all_results.get((n_qubits, max_depth_for_n), set())

        # More advanced: Canonicalize solutions to count families
        canonical_solutions = set()
        for sol in solutions:
            arr = np.array(sol)
            # Find all cyclic shifts of the solution
            shifts = {tuple(np.roll(arr, i)) for i in range(len(arr))}
            # Add the lexicographically smallest shift as the canonical representative
            canonical_solutions.add(min(shifts))

        num_solutions_per_n[n_qubits] = len(canonical_solutions)
        print(f"N = {n_qubits} (D={max_depth_for_n}): Found {len(solutions)} total solutions, representing {len(canonical_solutions)} unique families.")

    # Prepare data for plotting and fitting
    n_values = np.array(list(num_solutions_per_n.keys()))
    solution_counts = np.array(list(num_solutions_per_n.values()))

    # Plot the results
    plt.figure(figsize=(10, 6))
    plt.plot(n_values, solution_counts, 'o-', label="Discovered Solution Families")
    plt.xlabel("Number of Qubits (N)")
    plt.ylabel("Number of Unique Solution Families")
    plt.title("Relationship Between System Size and Emergent Complexity")
    plt.grid(True)
    plt.xticks(n_values)

    # Try to fit a simple predictive model (e.g., linear)
    # This will likely fail for complex relationships, but it's a start
    if len(n_values) > 1:
        try:
            # Fit a line: y = mx + c
            coeffs = np.polyfit(n_values, solution_counts, 1)
            p = np.poly1d(coeffs)
            plt.plot(n_values, p(n_values), "r--", label=f"Linear Fit: y={coeffs[0]:.2f}x + {coeffs[1]:.2f}")

            # Predict for N=10
            predicted_for_10 = p(10)
            print(f"\nLinear model predicts {predicted_for_10:.2f} solution families for N=10.")

        except Exception as e:
            print(f"\nCould not perform curve fitting: {e}")

    plt.legend()
    plt.show()
````

````python

# ==============================================================================
#  PREAMBLE: IMPORTS AND SETUP
# ==============================================================================
import numpy as np
import matplotlib.pyplot as plt
import time
from typing import Dict, List, Tuple, Set

# Qiskit Imports
import qiskit
from qiskit import QuantumCircuit
from qiskit.circuit import ParameterVector
from qiskit.circuit.library import CXGate, CZGate, SwapGate, IGate, XGate
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator

print("--- QCA N=9 Universe Exploration ---")
print(f"Qiskit version: {qiskit.__version__}")

# ==============================================================================
#  CORE COMPONENTS (Encapsulated and unchanged)
# ==============================================================================

GATES = {"CX": CXGate(), "CZ": CZGate(), "SWAP": SwapGate(), "I": IGate()}
class QuantumCellularAutomaton:
    def __init__(self, num_qunodes: int, interaction_config: Dict):
        self.num_qunodes = num_qunodes
        self.interaction_config = interaction_config
        self.simulator = AerSimulator(method='statevector')
    def _get_interaction_unitary(self, t1, t2): return self.interaction_config.get(tuple(sorted((t1, t2))), GATES["I"])
    def evolve_state(self, initial_state: np.ndarray) -> np.ndarray:
        if self.num_qunodes == 0: return np.array([])
        qc = QuantumCircuit(self.num_qunodes)
        for i, bit in enumerate(initial_state):
            if bit == 1: qc.x(i)
        if self.num_qunodes > 0: qc.barrier()
        for i in range(self.num_qunodes):
            j = (i + 1) % self.num_qunodes
            op = self._get_interaction_unitary(initial_state[i], initial_state[j])
            if self.num_qunodes > 1 and op.name != 'id':
                qc.append(op, [i, j])
        sv = Statevector(qc)
        probs = sv.probabilities_dict()
        max_prob_str = max(probs, key=probs.get)
        return np.array([int(bit) for bit in max_prob_str[::-1]])

def create_generator_qnn(n_q: int, d: int) -> QuantumCircuit:
    if n_q == 0: return QuantumCircuit(0)
    theta = ParameterVector('θ', n_q * d)
    qc = QuantumCircuit(n_q)
    for i in range(d):
        for j in range(n_q): qc.ry(theta[i * n_q + j], j)
        if n_q > 1:
            for j in range(n_q - 1): qc.cz(j, j + 1)
    qc.measure_all()
    return qc

def create_cost_function(qca: QuantumCellularAutomaton, qnn: QuantumCircuit, steps: int, shift: int):
    n_q = qnn.num_qubits
    if n_q == 0: return lambda theta: (1, [])
    backend = AerSimulator()
    def cost_func(theta):
        bound_qc = qnn.assign_parameters(theta)
        res = backend.run(bound_qc, shots=1, memory=True).result()
        s_init_str = res.get_memory()[0]
        s_init = np.array([int(b) for b in s_init_str[::-1]])
        activity = np.sum(s_init)
        if activity == 0 or activity == n_q: return n_q, s_init
        s_final = s_init
        for _ in range(steps): s_final = qca.evolve_state(s_final)
        s_target = np.roll(s_init, shift)
        cost = np.sum(s_final != s_target)
        return cost, s_init
    return cost_func

# ==============================================================================
#  THE META-EXPERIMENT: A FUNCTION TO RUN ONE FULL DISCOVERY
# ==============================================================================

def find_solutions_for_config(
    N: int, D: int, S: int, G: int,
    max_iter: int, num_runs: int, rules: Dict
) -> Set[Tuple[int, ...]]:
    print(f"\n--- Testing N={N}, D={D}, S={S}, G={G} ---")

    qca_universe = QuantumCellularAutomaton(num_qunodes=N, interaction_config=rules)
    generator_qnn = create_generator_qnn(n_q=N, d=D)
    cost_function = create_cost_function(qca=qca_universe, qnn=generator_qnn, steps=S, shift=G)

    unique_solutions_found = set()
    run_log_interval = num_runs // 10 if num_runs > 10 else 1

    for i in range(num_runs):
        if (i + 1) % run_log_interval == 0:
             print(f"  Run {i+1}/{num_runs}...", end='\r')

        best_cost_in_run = float('inf')
        for _ in range(max_iter):
            params = np.random.uniform(0, 2 * np.pi, generator_qnn.num_parameters)
            cost, state = cost_function(params)

            if cost < best_cost_in_run:
                best_cost_in_run = cost

            if cost == 0:
                unique_solutions_found.add(tuple(state))
                # Let's not break early to find more diverse solutions if they exist

    print(f"\nCompleted scan for N={N}, D={D}. Found {len(unique_solutions_found)} unique solutions in {num_runs} runs.")
    return unique_solutions_found

# ==============================================================================
#  THE MAIN CONTROL SCRIPT
# ==============================================================================

if __name__ == "__main__":
    # --- Define the Scan Parameters for the N=9 Test ---
    N_TO_TEST = 9
    QNN_DEPTH = 12          # Use a deep QNN to give it the best chance

    QCA_SIM_STEPS = 4
    GLIDER_SHIFT_TEMPLATE = -4

    RUNS_PER_CONFIG = 200   # Be thorough
    ITER_PER_RUN = 1000     # A deep search for each run

    # --- Define the "Laws of Physics" ---
    QCA_RULES = {
        (0, 0): IGate(),
        (1, 1): CXGate(),
        (0, 1): SwapGate()
    }

    # --- Data Collection ---
    scan_start_time = time.time()

    n_qubits = N_TO_TEST
    glider_shift = GLIDER_SHIFT_TEMPLATE % n_qubits if n_qubits > 0 else 0

    solutions = find_solutions_for_config(
        N=n_qubits, D=QNN_DEPTH, S=QCA_SIM_STEPS, G=glider_shift,
        max_iter=ITER_PER_RUN, num_runs=RUNS_PER_CONFIG, rules=QCA_RULES
    )

    scan_end_time = time.time()
    print(f"\n--- N=9 EXPLORATION COMPLETE IN {scan_end_time - scan_start_time:.2f} SECONDS ---")

    # --- ANALYSIS OF THE N=9 RESULT ---
    print("\n--- Analysis of N=9 Results ---")

    canonical_solutions = set()
    if solutions:
        for sol in solutions:
            arr = np.array(sol)
            shifts = {tuple(np.roll(arr, i)) for i in range(len(arr))}
            canonical_solutions.add(min(shifts))

    num_families = len(canonical_solutions)
    print(f"For N = {n_qubits}, we discovered {num_families} unique solution families.")

    print("\n--- Conclusion ---")
    if num_families == 0:
        print("Result: The N=9 universe appears to be 'sterile' for this type of life.")
        print("This strongly supports the hypothesis that odd-numbered systems are less complex.")
    elif num_families <= 3: # Assuming 3 is "low" compared to N=8's 5
        print(f"Result: The N=9 universe supports a low number of lifeforms ({num_families}).")
        print("This supports the hypothesis that odd-numbered systems are less complex than their even neighbors.")
    else:
        print(f"Result: A surprisingly high number of lifeforms ({num_families}) were found!")
        print("This REJECTS our simple 'even/odd' hypothesis and points to a deeper, more complex rule.")

    if canonical_solutions:
        print("\nDiscovered Solution Families:")
        for i, sol_family in enumerate(sorted(list(canonical_solutions))):
            print(f"  Family {i+1}: {np.array(sol_family)}")
````
